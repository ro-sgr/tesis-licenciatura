```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import axes3d
from mpl_toolkits.mplot3d.art3d import Line3DCollection
import numpy as np
from scipy.special import assoc_laguerre
from scipy.special import sph_harm_y # (l, m, θ, ϕ)

from math import factorial
import os
```

```python
def R_nl(n: int, l:int, a0: float, r:float) -> float:
    ρ: float = 2*r / (n*a0)
    norm: float = np.sqrt( np.power(2/(n*a0), 3) * factorial(n-l-1)/(2*n*factorial(n+l)) )
    laguerre: float = assoc_laguerre(ρ, n-l-1, 2*l+1)
    return norm * np.exp(-ρ/2) * np.power(ρ, l) * laguerre
```

```python
def ϕ_nlm(n:int, l:int, m:int, a0:float, r:float, θ: float, φ: float) -> float:
    return R_nl(n, l, a0, r) * sph_harm_y(l, m, θ, φ)
```

**Recordatorio.**

La función de onda tiene unidades de

$$ [\Psi] = \text{distancia}^{-3/2} \,.$$

La probabilidad es adimensional. Por lo que, la *densidad de probabilidad* debe de tener unidades de

$$ [\rho] = [\,|\Psi|^2\,] = \text{distancia}^{-3} = \frac{1}{\text{volumen}} \,.$$

## Densidad de probabilidad

La densidad de probabilidad es

$$ \rho(\mathbf{r}) = |\psi(\mathbf{r})|^2 = |\psi(r, \theta, \varphi)|^2 $$

```python
def ρ_nlm(n:int, l:int, m:int, a0:float, r:float, θ: float, φ: float) -> float:
    """ Densidad de probabilidad
    (n,l,m) : número cuántico (principal, azimutal, magnético)
    a0 : radio de Bohr
    r : distancia radial
    θ : ángulo azimutal
    φ : ángulo polar
    """
    return np.abs(ϕ_nlm(n, l, m, a0, r, θ, φ))**2
```

```python
def ρ_mallado(n: int, l: int, m: int, a0: float = 1.0,
              R_data: list = [0, 2, 3],
              θ_data: list = [0, np.pi, 10],
              φ_data: list = [0, 2*np.pi, 10]):
    """ Mallado de densidad de probabilidad
    """
    rmin, rmax, numR = R_data
    θmin, θmax, numθ = θ_data
    φmin, φmax, numφ = φ_data
    
    r = np.linspace(rmin, rmax, numR)
    θ = np.linspace(θmin, θmax, numθ)
    φ = np.linspace(φmin, φmax, numφ)
    R, Theta, Phi = np.meshgrid(r, θ, φ)

    # Calculate density
    densidad = ρ_nlm(n, l, m, a0, R, Theta, Phi)

    # Convert to Cartesian coordinates
    X = R * np.sin(Theta) * np.cos(Phi)
    Y = R * np.sin(Theta) * np.sin(Phi)
    Z = R * np.cos(Theta)
    
    return X, Y, Z, densidad
```

## Densidades de prob. que no son densidades

Hay muchas gráficas de estos cortes transversales en internet, sin embargo si uno presta atención a ello lo que están graficando es

$$ r^2 |\psi(\mathbf{r})|^2 = r^2 \sin\theta |\psi(r, \theta, \varphi)|^2 \,.$$

El integrando de la probabilidad sobre todo el espacio, por alguna mística razón que desconozco.

```python
def ρ_nlm_mala(n:int, l:int, m:int, a0:float, r:float, θ: float, φ: float) -> float:
    """ Densidad de probabilidad
    (n,l,m) : número cuántico (principal, azimutal, magnético)
    a0 : radio de Bohr
    r : distancia radial
    θ : ángulo azimutal
    φ : ángulo polar
    """
    return np.square(r) * np.sin(θ) * np.square(np.abs(ϕ_nlm(n, l, m, a0, r, θ, φ)))
```

```python
def ρ_mallado_malo(n: int, l: int, m: int, a0: float = 1.0,
              R_data: list = [0, 2, 3],
              θ_data: list = [0, np.pi, 10],
              φ_data: list = [0, 2*np.pi, 10]):
    """ Mallado de densidad de probabilidad
    """
    rmin, rmax, numR = R_data
    θmin, θmax, numθ = θ_data
    φmin, φmax, numφ = φ_data
    
    r = np.linspace(rmin, rmax, numR)
    θ = np.linspace(θmin, θmax, numθ)
    φ = np.linspace(φmin, φmax, numφ)
    R, Theta, Phi = np.meshgrid(r, θ, φ)

    # Calculate density
    densidad = ρ_nlm_mala(n, l, m, a0, R, Theta, Phi)

    # Convert to Cartesian coordinates
    X = R * np.sin(Theta) * np.cos(Phi)
    Y = R * np.sin(Theta) * np.sin(Phi)
    Z = R * np.cos(Theta)
    
    return X, Y, Z, densidad
```

# Generar conjuntos de imgs

```python
def data_plot_radio(n: int, l: int, m: int, a0: float = 1.0,
                    Rmin: float = 0.5, Rmax: float = 3, Rstep: float = 0.1):
    """ Radio que cambia
    """
    X_data: list = []
    Y_data: list = []
    Z_data: list = []
    density_data: list = []
    for R in np.arange(Rmin, Rmax, Rstep):
        X, Y, Z, density = ρ_mallado(n, l, m, a0)
        X_data.append(X)
        Y_data.append(Y)
        Z_data.append(Z)
        density_data.append(density)
        
    return X_data, Y_data, Z_data, density_data
```

```python
def guardar_imgs(n:int, l:int, m:int, a0:float = 1, Rmin:float = 0.5, Rmax:float = 5, Rstep:float = 0.1):
    """ Guarda las imágenes generadas para diferentes radios
    """
    X_data, Y_data, Z_data, density_data = data_plot_radio(n, l, m, a0, Rmin, Rmax, Rstep)
    
    max_density = np.max(density_data)
    xmax, ymax, zmax = np.max(X_data), np.max(Y_data), np.max(Z_data) # valores máximos en X, Y, Z
    xpadd, ypadd, zpadd = xmax*0.1, ymax*0.1, zmax*0.1 # x, y, z espaciado para graficar
    xmin, xmax = -xmax-xpadd, xmax+xpadd
    ymin, ymax = -ymax-ypadd, ymax+ypadd
    zmin, zmax = -zmax-zpadd, zmax+zpadd
    
    frames: int = len(X_data)
    count: int = 0
    avance: float = 0.25 # procentaje de gráficas guardadas
    
    for i in range(frames):
        X, Y, Z, density = X_data[i], Y_data[i], Z_data[i], density_data[i]
        
        # Plot
        fig = plt.figure(figsize=(10, 8))
        ax = fig.add_subplot(111, projection='3d')
        ax.plot_surface(X, Y, Z, facecolors=plt.cm.viridis(density / max_density),
                        rstride=1, cstride=1, antialiased=False, shade=True)
        
        # Aesthetics
        ax.set_title(f"Hydrogen Orbital Angular Shape (n={n}, ℓ={l}, m={m})")
        ax.set_xlabel("x")
        ax.set_ylabel("y")
        ax.set_zlabel("z")
        ax.set_box_aspect([1, 1, 1])
    
        ax.set_xlim(xmin, xmax)
        ax.set_ylim(ymin, ymax)
        ax.set_zlim(zmin, zmax)
        
        # Save plot
        output_dir = f"imgs/H_({n},{l},{m})"
        filename = f"frame_{i:03d}.png"
        filepath = os.path.join(output_dir, filename)
        plt.savefig(filepath, dpi=150)
        plt.close(fig)  # Close the figure to save memory

        if count >= avance*frames:
            print(f"{int(avance*100)}% de archivos guardados")
            avance += 0.25
        count += 1
        
    print(f"Todos los archivos se han guardado en {output_dir}")
```
