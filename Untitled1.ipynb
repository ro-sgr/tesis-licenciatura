```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
from scipy.special import assoc_laguerre
from scipy.special import sph_harm_y # (l, m, θ, ϕ)

from math import factorial
```

```python
def R_nl(n: int, l:int, a0: float, r:float) -> float:
    ρ: float = 2*r / (n*a0)
    norm: float = np.sqrt( np.power(2/(n*a0), 3) * factorial(n-l-1)/(2*n*factorial(n+l)) )
    laguerre: float = assoc_laguerre(ρ, n-l-1, 2*l+1)
    return norm * np.exp(-ρ/2) * np.power(ρ, l) * laguerre
```

```python
def ϕ_nlm(n:int, l:int, m:int, a0:float, r:float, θ: float, φ: float) -> float:
    return R_nl(n, l, a0, r) * sph_harm_y(l, m, θ, φ)
```

```python
def ρ_nlm(n:int, l:int, m:int, a0:float, r:float, θ: float, φ: float) -> float:
    """ Densidad de probabilidad
    (n,l,m) : número cuántico (principal, azimutal, magnético)
    a0 : radio de Bohr
    r : distancia radial
    θ : ángulo azimutal
    φ : ángulo polar
    """
    return np.abs(ϕ_nlm(n, l, m, a0, r, θ, φ))**2
```

```python
def cartesian_to_spherical(x, y, z):
    r = np.sqrt(x**2 + y**2 + z**2)
    θ = np.arccos(np.divide(z, r, out=np.zeros_like(r), where=r!=0))  # polar angle
    φ = np.arctan2(y, x)  # azimuthal angle
    return r, θ, φ
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import sph_harm_y, factorial, assoc_laguerre

# Radial part function (from your code)
def R_nl(n, l, a0, r):
    ρ = 2 * r / (n * a0)
    norm = np.sqrt((2 / (n * a0))**3 * factorial(n - l - 1) / (2 * n * factorial(n + l)))
    laguerre = assoc_laguerre(ρ, n - l - 1, 2 * l + 1)
    return norm * np.exp(-ρ / 2) * ρ**l * laguerre

# Total wavefunction (complex)
def ψ_nlm(n, l, m, a0, r, θ, φ):
    R = R_nl(n, l, a0, r)
    Y = sph_harm_y(l, m, θ, φ)
    return R * Y
```

```python
# Parameters
a0 = 1  # Bohr radius in atomic units
n, l, m = 3, 2, 2
```

```python
# Grid in spherical coordinates
num_points = 100
theta = np.linspace(0, np.pi, num_points)  # polar angle
phi = np.linspace(0, 2 * np.pi, num_points)  # azimuthal angle
theta, phi = np.meshgrid(theta, phi)

# Radius r depends on the amplitude of wavefunction to get surface shape
# For visualization, scale wavefunction magnitude
r = 20 * np.abs(ψ_nlm(n, l, m, a0, 1, theta, phi))  # fix r=1 for radial slice, scale for visibility

# Convert spherical to Cartesian coordinates for plotting
X = r * np.sin(theta) * np.cos(phi)
Y = r * np.sin(theta) * np.sin(phi)
Z = r * np.cos(theta)

# Plotting
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

# Color by phase of wavefunction (real part sign)
colors = np.angle(ψ_nlm(n, l, m, a0, 1, theta, phi))
surf = ax.plot_surface(X, Y, Z, facecolors=plt.cm.hsv((colors + np.pi) / (2 * np.pi)), rstride=1, cstride=1, antialiased=False)


# Get the max radius value from X, Y, Z coordinates
Rmax = np.max(np.abs([X, Y, Z]))

# Set axis limits to match
ax.set_xlim(-Rmax, Rmax)
ax.set_ylim(-Rmax, Rmax)
ax.set_zlim(-Rmax, Rmax)

ax.set_box_aspect([1,1,1])
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title(f'Hydrogenic orbital ψ({n},{l},{m})')

plt.show()
```
